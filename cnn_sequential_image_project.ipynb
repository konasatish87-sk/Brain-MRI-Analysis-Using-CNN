```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow import keras
```

```python
training_path='/content/drive/MyDrive/archive (1)/Training'
testing_path='/content/drive/MyDrive/archive (1)/Testing'
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    training_path,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(224, 224),
    batch_size=32
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    training_path,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(224, 224),
    batch_size=32
)
```

```python
test_ds = tf.keras.utils.image_dataset_from_directory(
    testing_path,
    image_size=(224, 224),
    batch_size=32,
    shuffle=False
)
```

```python
print(train_ds.class_names)
print(len(train_ds.class_names))
```

```python
#building cnn model
from tensorflow.keras import layers,models
cnn_model=models.Sequential([
    layers.Input(shape=(244,244,3)),
    #converting pixel values form [0,255] to [0,1]
    layers.Rescaling(1.0/255),
    #cnn feature learning layer
    layers.Conv2D(32,(3,3),activation='relu'),
    layers.MaxPooling2D(),

    layers.Conv2D(64,(3,3),activation='relu'),
    layers.MaxPooling2D(),

     layers.Conv2D(128,(3,3),activation='relu'),
     layers.MaxPooling2D(),
    #classification layer
    layers.GlobalAveragePooling2D(),
    layers.Dense(128,activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(4,activation='softmax')
    ])
```

```python
cnn_model.summary
```

```python
cnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
```

```python
history=cnn_model.fit(train_ds,
                      validation_data=val_ds,epochs=10)
```

```python
cnn_loss,cnn_accuracy=cnn_model.evaluate(val_ds)
print(f'cnn from scratch-validation accuracy:{cnn_accuracy:.3f}')
```

```python
cnn_model.save('cnn_simple_from_scratch.keras')
```
